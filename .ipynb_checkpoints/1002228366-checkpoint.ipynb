```python
import cv2
import matplotlib.pyplot as plt
import numpy as np

img = cv2.imread("C:\\Users\\varsh\\Downloads\\CV_MASTERS\\PA1-1\\PA1\\Images\\rgb.png")
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)


plt.imshow(img_rgb)
plt.title("Original RGB Image")
plt.axis('off')
plt.show()


R = img_rgb[:, :, 0]
G = img_rgb[:, :, 1]
B = img_rgb[:, :, 2]

#a
fig, axs = plt.subplots(1, 3, figsize=(15, 5))
axs[0].imshow(R, cmap='gray'); axs[0].set_title('Red Channel'); axs[0].axis('off')
axs[1].imshow(G, cmap='gray'); axs[1].set_title('Green Channel'); axs[1].axis('off')
axs[2].imshow(B, cmap='gray'); axs[2].set_title('Blue Channel'); axs[2].axis('off')
plt.tight_layout()
plt.show()


#b
gray = (0.33 * R + 0.33 * G + 0.33 * B).astype(np.uint8)

plt.imshow(gray, cmap='gray')
plt.title("Manual Grayscale Image")
plt.axis('off')
plt.show()


#c
red_mask = (R > 150) & (G < 100) & (B < 100)
only_red = np.zeros_like(img_rgb)
only_red[red_mask] = img_rgb[red_mask]


plt.imshow(only_red)
plt.title("Only Red Pixels")
plt.axis('off')
plt.show()


#d
img_hsv = cv2.cvtColor(img_rgb, cv2.COLOR_RGB2HSV)
hsv_modified = img_hsv.copy()
hsv_modified[:, :, 2] = (hsv_modified[:, :, 2] * 0.5).astype(np.uint8)
darker_rgb = cv2.cvtColor(hsv_modified, cv2.COLOR_HSV2RGB)


plt.imshow(darker_rgb)
plt.title("RGB Image with 50% Brightness")
plt.axis('off')
plt.show()



```

```python
import cv2
import numpy as np 
import matplotlib.pyplot as plt
apple_bgr = cv2.imread(r"C:\Users\varsh\Downloads\CV_MASTERS\PA1-1\PA1\Images\apple.jpg")
apple_rgb = cv2.cvtColor(apple_bgr, cv2.COLOR_BGR2RGB)

R = apple_rgb[:, :, 0]
G = apple_rgb[:, :, 1]
B = apple_rgb[:, :, 2]


rgb_mask = (G > 80) & (R < 130) & (B < 130)
apple_rgb_masked = np.zeros_like(apple_rgb)
apple_rgb_masked[rgb_mask] = apple_rgb[rgb_mask]


apple_hsv = cv2.cvtColor(apple_rgb, cv2.COLOR_RGB2HSV)


lower_green = np.array([35, 50, 50])
upper_green = np.array([85, 255, 255])

hsv_mask = cv2.inRange(apple_hsv, lower_green, upper_green)
apple_hsv_masked = cv2.bitwise_and(apple_rgb, apple_rgb, mask=hsv_mask)


fig, axs = plt.subplots(1, 3, figsize=(18, 6))

axs[0].imshow(apple_rgb)
axs[0].set_title("Original Apple")
axs[0].axis('off')

axs[1].imshow(apple_rgb_masked)
axs[1].set_title("RGB Masked")
axs[1].axis('off')

axs[2].imshow(apple_hsv_masked)
axs[2].set_title("HSV Masked")
axs[2].axis('off')

plt.tight_layout()
plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
bags_bgr = cv2.imread(r"C:\Users\varsh\Downloads\CV_MASTERS\PA1-1\PA1\Images\bags.png")
bags_rgb = cv2.cvtColor(bags_bgr, cv2.COLOR_BGR2RGB)
R = bags_rgb[:, :, 0]
G = bags_rgb[:, :, 1]
B = bags_rgb[:, :, 2]

rgb_mask = (
    (R >= 170) &        
    (G >= 60) & (G <= 120) &  
    (B <= 70)            
)

bags_rgb_masked = np.zeros_like(bags_rgb)
bags_rgb_masked[rgb_mask] = bags_rgb[rgb_mask]
orange_hsv = cv2.cvtColor(bags_rgb_masked, cv2.COLOR_RGB2HSV)
orange_hsv[:, :, 1] = np.clip(orange_hsv[:, :, 1] * 1.8, 0, 255).astype(np.uint8)
enhanced_orange = cv2.cvtColor(orange_hsv, cv2.COLOR_HSV2RGB)
fig, axs = plt.subplots(1, 3, figsize=(18, 6))

axs[0].imshow(bags_rgb)
axs[0].set_title("Original Bags Image")
axs[0].axis('off')

axs[1].imshow(bags_rgb_masked)
axs[1].set_title("RGB Masked (Orange Only)")
axs[1].axis('off')

axs[2].imshow(enhanced_orange)
axs[2].set_title("Enhanced Orange Bag")
axs[2].axis('off')

plt.tight_layout()
plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
pumpkins_bgr = cv2.imread(r"C:\Users\varsh\Downloads\CV_MASTERS\PA1-1\PA1\Images\pumpkins.jpeg")
pumpkins_rgb = cv2.cvtColor(pumpkins_bgr, cv2.COLOR_BGR2RGB)
pumpkins_float = pumpkins_rgb.astype(np.float32) / 255.0
# a
darker = np.clip(pumpkins_rgb.astype(np.int16) - 128, 0, 255).astype(np.uint8)

# b
lower_contrast = (np.power(pumpkins_float, 0.25) * 255).astype(np.uint8)

# c
inverted = cv2.bitwise_not(pumpkins_rgb)  

# d
brighter = np.clip(pumpkins_rgb.astype(np.int16) + 128, 0, 255).astype(np.uint8)

# e
higher_contrast = (np.power(pumpkins_float, 3.0) * 255).astype(np.uint8)
fig, axs = plt.subplots(2, 3, figsize=(15, 10))

axs[0, 0].imshow(pumpkins_rgb)
axs[0, 0].set_title("Original Image")
axs[0, 0].axis('off')

axs[0, 1].imshow(darker)
axs[0, 1].set_title("Darker Image ")
axs[0, 1].axis('off')

axs[0, 2].imshow(lower_contrast)
axs[0, 2].set_title("Lower Contrast")
axs[0, 2].axis('off')

axs[1, 0].imshow(inverted)
axs[1, 0].set_title("Inverted Image")
axs[1, 0].axis('off')

axs[1, 1].imshow(brighter)
axs[1, 1].set_title("Brighter Image ")
axs[1, 1].axis('off')

axs[1, 2].imshow(higher_contrast)
axs[1, 2].set_title("Higher Contrast")
axs[1, 2].axis('off')

plt.tight_layout()
plt.show()
```

TASK 4 

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
import random
img_path = r'C:\Users\varsh\Downloads\CV_MASTERS\PA1-1\PA1\Images\cameraman.jpeg'
img = cv2.imread(img_path, cv2.IMREAD_GRAYSCALE)

# a
blur_5 = cv2.blur(img, (5, 5))
blur_11 = cv2.blur(img, (11, 11))
plt.figure(figsize=(12, 4))
for i, (title, image) in enumerate([
    ('Original', img),
    ('Avg Blur 5x5', blur_5),
    ('Avg Blur 11x11', blur_11)
]):
    plt.subplot(1, 3, i + 1)
    plt.imshow(image, cmap='gray')
    plt.title(title)
    plt.axis('off')
plt.tight_layout()
plt.show()

# b

def add_salt_pepper_noise(image, amount):
    noisy = image.copy()
    total = int(amount * image.size)
    coords = [np.random.randint(0, i - 1, total) for i in image.shape]
    for i in range(total):
        if random.random() < 0.5:
            noisy[coords[0][i], coords[1][i]] = 0   
        else:
            noisy[coords[0][i], coords[1][i]] = 255 
    return noisy

img_sp10 = add_salt_pepper_noise(img, 0.10)
gauss_blur_5 = cv2.GaussianBlur(img_sp10, (5, 5), 2.0)
gauss_blur_11 = cv2.GaussianBlur(img_sp10, (11, 11), 5.0)

plt.figure(figsize=(12, 4))
for i, (title, image) in enumerate([
    ('Salt & Pepper 10%', img_sp10),
    ('Gaussian Blur 5x5 σ=2', gauss_blur_5),
    ('Gaussian Blur 11x11 σ=5', gauss_blur_11)
]):
    plt.subplot(1, 3, i + 1)
    plt.imshow(image, cmap='gray')
    plt.title(title)
    plt.axis('off')
plt.tight_layout()
plt.show()

# c

img_sp20 = add_salt_pepper_noise(img, 0.20)
med_blur_5 = cv2.medianBlur(img_sp20, 5)
med_blur_11 = cv2.medianBlur(img_sp20, 11)

plt.figure(figsize=(12, 4))
for i, (title, image) in enumerate([
    ('Salt & Pepper 20%', img_sp20),
    ('Median Blur 5x5', med_blur_5),
    ('Median Blur 11x11', med_blur_11)
]):
    plt.subplot(1, 3, i + 1)
    plt.imshow(image, cmap='gray')
    plt.title(title)
    plt.axis('off')
plt.tight_layout()
plt.show()

# d

gauss_blur_7 = cv2.GaussianBlur(img, (7, 7), 4.0)
img_resized = cv2.resize(img, (50, 50), interpolation=cv2.INTER_AREA)
blur_resized = cv2.resize(gauss_blur_7, (50, 50), interpolation=cv2.INTER_AREA)

plt.figure(figsize=(12, 4))
for i, (title, image) in enumerate([
    ('Original', img),
    ('Original 50x50', img_resized),
    ('Blurred 50x50', blur_resized)
]):
    plt.subplot(1, 3, i + 1)
    plt.imshow(image, cmap='gray')
    plt.title(title)
    plt.axis('off')
plt.tight_layout()
plt.show()

# e

G0 = img
G1 = cv2.pyrDown(G0)
G2 = cv2.pyrDown(G1)

plt.figure(figsize=(12, 4))
for i, (title, image) in enumerate([
    ('Level 0', G0),
    ('Level 1', G1),
    ('Level 2', G2)
]):
    plt.subplot(1, 3, i + 1)
    plt.imshow(image, cmap='gray')
    plt.title(f'Gaussian {title}')
    plt.axis('off')
plt.tight_layout()
plt.show()

# f

L0 = cv2.subtract(G0, cv2.pyrUp(G1, dstsize=G0.shape[::-1]))
L1 = cv2.subtract(G1, cv2.pyrUp(G2, dstsize=G1.shape[::-1]))
L2 = G2.copy()  

plt.figure(figsize=(12, 4))
for i, (title, image) in enumerate([
    ('Level 0', L0),
    ('Level 1', L1),
    ('Level 2', L2)
]):
    plt.subplot(1, 3, i + 1)
    plt.imshow(image, cmap='gray')
    plt.title(f'Laplacian {title}')
    plt.axis('off')
plt.tight_layout()
plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# a
cmu = cv2.imread(r"C:\Users\varsh\Downloads\CV_MASTERS\PA1-1\PA1\Images\cmu.jpg")
cmu_rgb = cv2.cvtColor(cmu, cv2.COLOR_BGR2RGB)
blurred_cmu = cv2.GaussianBlur(cmu_rgb, (25, 25), 15)
shifted_shadow = np.roll(blurred_cmu, shift=(10, 10), axis=(0, 1))  
shadow_effect = np.maximum(shifted_shadow, cmu_rgb)

# b
dallas = cv2.imread(r"C:\Users\varsh\Downloads\CV_MASTERS\PA1-1\PA1\Images\dallas.jpg")
dallas_rgb = cv2.cvtColor(dallas, cv2.COLOR_BGR2RGB)
blurred_dallas = cv2.GaussianBlur(dallas_rgb, (25, 25), 15)
tilt_shift = blurred_dallas.copy()
tilt_shift[250:350] = dallas_rgb[250:350]
fig, axs = plt.subplots(1, 2, figsize=(14, 6))

axs[0].imshow(shadow_effect)
axs[0].set_title("Task (a): Shadow Effect on CMU")
axs[0].axis('off')

axs[1].imshow(tilt_shift)
axs[1].set_title("Task (b): Tilt-Shift Effect on dallas")
axs[1].axis('off')

plt.tight_layout()
plt.show()
```
