```python
import pandas as pd 

import polars as pl 

import time
```

```python
import csv, random, datetime


#makes the random number generator start from a fixed point every time. 
#So whenever you run the data generation cell, you get the exact same 10,000 rows — same products, same nulls, same prices.

random.seed(42) 
products = ['laptop', 'LAPTOP', 'Phone', 'phone', 'Tablet', 'tablet', 'monitor', 'Monitor']
regions  = ['North', 'South', 'East', 'West', None]
statuses = ['completed', 'pending', 'Cancelled', 'completed', 'completed', 'pending']

rows = [['order_id','product','quantity','price','region','status','order_date']]
for i in range(1, 10001):
    product = random.choice(products)
    qty     = random.randint(1, 20) if random.random() > 0.03 else None
    price   = round(random.uniform(100, 3000), 2) if random.random() > 0.02 else None
    region  = random.choice(regions)
    status  = random.choice(statuses)
    days    = random.randint(0, 364)
    date    = (datetime.date(2024, 1, 1) + datetime.timedelta(days=days)).isoformat()
    rows.append([i, product, qty, price, region, status, date])

with open('sales.csv', 'w', newline='') as f:
    csv.writer(f).writerows(rows)

print("Done! sales.csv created with 10,000 rows")
```

```python
# pandas
t0 = time.perf_counter()
pdf = pd.read_csv('sales.csv')
print(f"pandas load: {(time.perf_counter()-t0)*1000:.1f}ms")
print(pdf.head())
```

```python
# polars
t0 = time.perf_counter()
pldf = pl.read_csv('sales.csv')
print(f"polars load: {(time.perf_counter()-t0)*1000:.1f}ms")
print(pldf.head())
```

```python
# The timer stops before head() runs. 
# So you're measuring how fast each library reads and loads the entire 10,000 row CSV into memory.
# Polars is faster at that because:

# pandas reads row by row into NumPy arrays
# polars reads column by column using Apache Arrow format, which is much more efficient for columnar data
# polars is written in Rust under the hood, which is faster than pandas' Python/C mix
```

```python
# pandas
t0 = time.perf_counter()
pdf["product"] = pdf["product"].str.lower().str.strip()
pandas_str = (time.perf_counter()-t0)*1000
print(pdf["product"].unique())
```

```python
# polars
pldf = pldf.with_columns(
    pl.col("product").str.to_lowercase().str.strip_chars()
)
print(pldf["product"].unique())
```

```python
# pandas
pdf_clean = pdf.dropna(subset=["price", "quantity"])
print(f"pandas: {len(pdf)} → {len(pdf_clean)} rows after dropping nulls")
print(pdf_clean[["quantity","price"]].dtypes)
```

```python
# polars
pldf_clean = pldf.drop_nulls(subset=["price", "quantity"])
print(f"polars: {len(pldf)} → {len(pldf_clean)} rows after dropping nulls")
print(pldf_clean[["quantity","price"]].dtypes)
```

```python
# pandas — need .copy() first to avoid SettingWithCopyWarning
pdf_clean = pdf_clean.copy()
pdf_clean["revenue"] = pdf_clean["quantity"] * pdf_clean["price"]
print(pdf_clean[["quantity", "price", "revenue"]].head(3))
```

```python
# polars
pldf_clean = pldf_clean.with_columns(
    (pl.col("quantity") * pl.col("price")).alias("revenue")
)
print(pldf_clean[["quantity", "price", "revenue"]].head(3))
```

```python
# pandas
pdf_completed = pdf_clean[pdf_clean["status"] == "completed"]
print(f"pandas: {len(pdf_completed)} completed orders")
```

```python
# polars
pldf_completed = pldf_clean.filter(pl.col("status") == "completed")
print(f"polars: {len(pldf_completed)} completed orders")
```

```python
# pandas
t0 = time.perf_counter()
pdf_grp = (
    pdf_completed
    .groupby("product")
    .agg(
        total_revenue=("revenue", "sum"),
        avg_price=("price", "mean"),
        orders=("order_id", "count")
    )
    .sort_values("total_revenue", ascending=False)
)
print(f"pandas groupby: {(time.perf_counter()-t0)*1000:.2f}ms")
print(pdf_grp)
```

```python
# polars
t0 = time.perf_counter()
pldf_grp = (
    pldf_completed
    .group_by("product")
    .agg([
        pl.col("revenue").sum().alias("total_revenue"),
        pl.col("price").mean().alias("avg_price"),
        pl.col("order_id").count().alias("orders"),
    ])
    .sort("total_revenue", descending=True)
)
print(f"polars groupby: {(time.perf_counter()-t0)*1000:.2f}ms")
print(pldf_grp)
```

```python
t0 = time.perf_counter()

result = (
    pl.scan_csv("sales.csv")          # lazy — nothing runs yet
    .with_columns(
        pl.col("product").str.to_lowercase().str.strip_chars()
    )
    .drop_nulls(subset=["price", "quantity"])
    .with_columns(
        (pl.col("quantity") * pl.col("price")).alias("revenue")
    )
    .filter(pl.col("status") == "completed")
    .group_by("product")
    .agg([
        pl.col("revenue").sum().alias("total_revenue"),
        pl.col("price").mean().alias("avg_price"),
        pl.col("order_id").count().alias("orders"),
    ])
    .sort("total_revenue", descending=True)
    .collect()                         # NOW it actually runs
)

print(f"polars lazy full pipeline: {(time.perf_counter()-t0)*1000:.2f}ms")
print(result)
```

```python
import matplotlib.pyplot as plt

steps = ['Load', 'String Clean', 'Drop Nulls', 'New Column', 'Filter', 'GroupBy']
pandas_times = [48.6, 3.45, 2.22, 1.32, 1.51, 22.56]
polars_times = [21.8, 2.17, 0.96, 0.31, 0.64, 6.43]
```

```python
x = range(len(steps))
width = 0.35

fig, ax = plt.subplots(figsize=(10, 5))
ax.bar([i - width/2 for i in x], pandas_times, width, label='pandas', color='blue')
ax.bar([i + width/2 for i in x], polars_times, width, label='polars', color='green')
ax.set_ylabel('Time (ms)')
ax.set_title('pandas vs polars — time per operation (lower is better)')
ax.set_xticks(list(x))
ax.set_xticklabels(steps)
ax.legend()
plt.tight_layout()
plt.show()
```
