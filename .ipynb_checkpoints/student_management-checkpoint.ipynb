```python
#### students = {}

def add_student():
    Name = input("Enter student Name:")
    roll_number = input("Enter Roll Number:")
    
    if roll_number in students:
        print("Roll number already exists, Please enter new roll number.")
        return 
    students[roll_number] = {"Name":Name, 
                             "marks":[], 
                             "attendance":set()
                            }
    print("student added sucessfully")
    

def update_marks():
    roll_number = input("Enter Roll Number: ")
    
    if roll_number in students:
        # Use input() to get marks from the user
        marks_input = input("Enter marks (comma-separated): ")
        
        # Split the marks string into a list and convert each mark to an integer
        marks = [int(mark.strip()) for mark in marks_input.split(",")]
        
        # Update the student's marks
        students[roll_number]["marks"] = marks
        
        # Print the updated marks
        print(f"Marks updated successfully! Marks: {students[roll_number]['marks']}")
        
        # Calculate and print the average marks
        avg_marks = sum(students[roll_number]["marks"]) / len(students[roll_number]["marks"])
        print(f"Average Marks: {avg_marks:.2f}")
        
    else:
        print("Roll Number not found.")

 
from datetime import datetime
def mark_attendance():
    roll_number = input("Enter Roll Number")
    if roll_number in students:
        date = input("Enter date (YYYY-MM-DD):").strip()
        students[roll_number]["attendance"].add(date)
        print("attendance recorded sucessfully!")
    else:
        print("Roll Number not found")

        
def view_student_details():
    print("Current students dictionary:", students)  # Print the entire dictionary before viewing details
    roll_number = input("Enter roll number to view details: ")
    
    if roll_number in students:
        student = students[roll_number]  # Access the student record using roll number
        print(f"Student Name: {student['Name']}")  # Access the student's Name correctly
        print(f"Roll Number: {roll_number}")  # Roll number is already the key
        print(f"Marks: {student['marks']}")  # Access the student's marks
        print(f"Attendance Dates: {sorted(student['attendance'])}")  # Access attendance dates
    else:
        print("Roll Number not found.")


def delete_student():
    roll_number = input("Enter Roll Number to delete: ")
    
    if roll_number in students:
        del students[roll_number]  # Delete the student record from the dictionary
        print(f"Student with Roll Number {roll_number} has been deleted successfully.")
    else:
        print("Roll Number not found.")




        
add_student()
update_marks()
mark_attendance()
view_student_details()
delete_student()
```
