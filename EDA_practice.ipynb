```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
data = pd.read_csv(r"C:\Users\varsh\OneDrive\Desktop\jupyter_practice\Employee_dataset.csv")
print(data.head())
```

```python
sns.histplot(data["age"])
```

```python
sns.countplot(data=data, x="gender_full")
```

```python
x = data["STATUS_YEAR"].value_counts()
plt.pie(x.values, labels = x.index, autopct = "%1.1f%%")
plt.show()
```

```python
plt.figure(figsize=(15,5))
sns.barplot(x=data["department_name"], y=data["length_of_service"])
plt.xticks(rotation=90)
```

```python
sns.scatterplot(x=data["length_of_service"],y=data["age"])
```

```python
sns.countplot(x="STATUS_YEAR",hue ="STATUS",data=data)
```

```python
from sklearn import datasets, decomposition
iris = datasets.load_iris()
X = iris.data
y = iris.target
pca = decomposition.PCA(n_components=2)
X = pca.fit_transform(X)
sns.scatterplot(x=X[:, 0], y=X[:, 1], hue=y)
```

```python
sns.heatmap(data.select_dtypes(include=['number']).corr(), annot=True)
```
